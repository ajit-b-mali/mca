```python
class Pet:
    def __init__(self, petType, name, price):
        self.name = name
        self.type = petType
        self.price = price

        
    def display(self):
        print(f"Type: {self.type}, Name: {self.name}, Price: {self.price}")

myPet = Pet("cat", "happy", "1250")
myPet.display()
```
