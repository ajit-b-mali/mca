```python
import re

def evaluteReview(review):
    review = review.lower()
    review = re.split(r"\W+", review)
    
    posReviewWords = ["good", "great"]
    negReviewWords = ["bad", "horrible"]
    
    for word in negReviewWords:
        if word in review:
            return "negative"
            
    for word in posReviewWords:
        if word in review:
            return "positive"
            
    return "neutral"


review = input("Enter your review to test:")
print(f"The review is {evaluteReview(review)}")
```
