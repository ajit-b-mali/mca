```python
tLog = input("Enter transaction log: ")
tLogList = list(map(lambda x: x.split(" "), tLog.split(", ")))

balance = 0

for tEntry in tLogList:
    tType = tEntry[0]
    tAmount = int(tEntry[1])
    if tType == "D":
        balance += tAmount
    elif tType == "W":
        if balance >= 500:
            balance -= tAmount

print("Current balance in account:", balance)
```
