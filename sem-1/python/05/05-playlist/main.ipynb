```python
import random

def menu():
    choice = int(input("\nSelect menu: "))
    if choice < 0 or choice > 5:
        print("\nInvalid choice Try again\n")
        menu()
    else:
        return choice


def addSong(playList: list):
    songName = input("Which song: ").lower()
    if songName in playList:
        print("Song already exist")
        return
    if songName:
        playList.append(songName)
    print("Updated playlist:", playList)


def playSpecific(playList: list):
    songName = input("Which song: ").lower()
    if songName in playList:
        print("Playing ", songName, "...")
    else:
        print("Error:", songName, "not found")


def removeSpecific(playList: list):
    songName = input("Which song: ").lower()
    if songName in playList:
        playList.remove(songName)
        print("Updated playlist:", playList)
    else:
        print("Song not found")


def shufflePlayList(playList: list):
    print("Before", playList)
    random.shuffle(playList)
    print("After", playList)

    
def playAll(playList: list):
    for song in playList:
        print("Playing", song, "....")
    

playList = ["aadat", "bada pachtaoge", "tadap tadap", "dj", "begging", "koyi nahi"]

print("This is menu")
print("1: add new song")
print("2: play specific song")
print("3: remove a song")
print("4: shuffle order")
print("5: play all")
print("0: exit")

print("\nCurrent songs:", playList)

while True:
    choice = menu()
    if choice == 0:
        print("BYEeee...")
        break
        
    elif choice == 1:
        addSong(playList)
        
    elif choice == 2:
        playSpecific(playList)
            
    elif choice == 3:
        removeSpecific(playList)
        
    elif choice == 4:
        shufflePlayList(playList)
        
    elif choice == 5:
        playAll(playList)

    else:
        print("Error: Wrong choice")
```
