```python
def deleteDuplicates(library):
    seen = set()
    unique_library = list()
    for book in library:
        bookName = book["title"]
        if bookName not in seen:
            seen.add(bookName)
            unique_library.append(book)
    
    return unique_library

def sortByCost(library):
    return sorted(library, key=lambda l: l['cost'])

def count500(library):
    cnt = 0;
    for book in library:
        if book['cost'] == 500:
            cnt += 1;
    return cnt

def printLibrary(msg, library):
    print(msg)
    for book in library:
        print(f"name = {book['title']} ---- cost = {book['cost']}")

library = [
    {"title": "Programming in Python", "author": "Dr. Rizwan rehman", "cost": 500},
    {"title": "Data Structures and Algorithms", "author": "Prof. Naveen Garg", "cost": 300},
    {"title": "Programming in Python", "author": "Dr. Rizwan rehman", "cost": 500},
    {"title": "Internet of Things: A Hands-On Approach", "author": "Vijay Wadisetti", "cost": 700},
    {"title": "Research Methodology", "author": "Ranjit Kumar", "cost": 500},
    {"title": "Elements of Discrete Mathematics", "author": "C. L. Liu", "cost": 300},
    {"title": "Internet of Things: A Hands-On Approach", "author": "Vijay Wadisetti", "cost": 700},
    {"title": "Internet of Things: A Hands-On Approach", "author": "Vijay Wadisetti", "cost": 700},
]

printLibrary("Current Libray:", library)

# delete duplicates
library = deleteDuplicates(library)
printLibrary("\nDeleting Duplicates\nUpdated Library:", library)

# sorting
printLibrary("\nSorting by Cost\nSort library:", sortByCost(library))

# count books which has cost 500
print("\nCounting books\n...", count500(library), "books has cost of 500")
```
