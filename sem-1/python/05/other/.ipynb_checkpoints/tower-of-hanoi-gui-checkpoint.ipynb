```python
import tkinter as tk

class TowerOfHanoi:
    def __init__(self, root, diskCnt):
        self.root = root
        self.canvas = tk.Canvas(self.root, width=640, height=480)
        self.canvas.pack()
        self.diskCnt = diskCnt

        self.rods = [[], [], []]
        self.moves = []

        for i in range(3):
            x = 150 + i * 200
            self.canvas.create_line(x, 100, x, 250, width=5)

        for i in range(self.diskCnt, 0, -1):
            disk = self.createDisk(i)
            self.rods[0].append(disk)
            self.placeDisk(disk, 0)

        self.solve(self.diskCnt, 0, 1, 2)

        self.moveIndex = 0
        self.root.after(1000, self.animate)


    def createDisk(self, size):
        width = 20 + size * 20
        return self.canvas.create_rectangle(0, 0, width, 20, fill="blue")

    def placeDisk(self, disk, rodIndex):
        rodX = 150 + rodIndex * 200
        diskPos = self.canvas.coords(disk)
        w = diskPos[2] - diskPos[0]
        rod = self.rods[rodIndex]
        y = 250 - len(rod) * 20
        self.canvas.coords(disk, rodX - w // 2, y, rodX + w // 2, y + 20)

    def solve(self, n, source, using, target):
        if n <= 0:
            return

        self.solve(n - 1, source, target, using)
        self.moves.append((source, target))
        self.solve(n - 1, using, source, target)

    def moveDisk(self, source, target):
        disk = self.rods[source].pop()
        self.rods[target].append(disk)
        self.placeDisk(disk, target)

    def animate(self):
        if self.moveIndex < len(self.moves):
            source, target = self.moves[self.moveIndex]
            self.moveDisk(source, target)
            self.moveIndex += 1
            self.root.after(250, self.animate)

if __name__ == "__main__":
    root = tk.Tk()
    root.title("Tower of Hanoi")
    toh = TowerOfHanoi(root, 3)
    root.mainloop()
```
