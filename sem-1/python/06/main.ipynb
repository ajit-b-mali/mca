```python
def linearSearch(arr, value):
    for i in range(len(arr)):
        if arr[i] == value:
            return i
    return

arr = [2, -1, 5, 10, 20, 12, -1, 50, -4, -6, 3]
print("Our List:", arr)
value = int(input("search what? "))

index = linearSearch(arr, value)
if not index:
    print(value, "not found")
else:
    print(f"{value} found at {index}th index")
```
