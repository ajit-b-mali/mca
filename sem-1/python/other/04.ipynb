```python
from getpass import getpass

balance = 0

def deposit():
    global balance
    temp = int(input("How much: "))
    if temp < 0:
        print("Can not add negative balance")
    else:
        balance += temp
        print("\ndone\n")


def withdraw():
    global balance
    temp = int(input("How much: "))
    if balance < 500:
        print("\nCan not withdraw money")
        print("Account should mentain minimum balance 500\n")
    elif temp > balance:
        print("\nNot enough balance\n")
    else:
        balance -= temp
        print("\ndone\n")
    

def checkBalance():
    print("\nbalance ->", balance, '\n')


def menu():
    print("Select a option")
    print("1: check balance")
    print("2: deposit")
    print("3: withdraw")
    print("0: exit")
    choice = int(input("choice: "))
    if choice < 0 or choice > 3:
        print("\nInvalid choice Try again\n")
        menu()
    else:
        return choice


def login():
    name = input("Enter your name: ")
    accountNumber = int(input("Enter account number: "))
    password = getpass("Enter password: ")
    
    if password != "00000000":
        print("wrong password")
        return ""
    
    return name


def main():
    name = login().upper()
    if not name:
        return
    
    print("\nLogged in as: ", name, '\n')
    while True:
        choice = menu()
        if choice == 0:
            print("Bye", name)
            break
        elif choice == 1:
            checkBalance()
        elif choice == 2:
            deposit()
        elif choice == 3:
            withdraw()
        else:
            print("Error: Wrong choice")
                

print("This is MCOE bank")
choice = input("Do you want to log in (y/n): ")
if choice == 'y' or choice == 'Y':
    main()

print("Exit")
```
