```python
num1 = int(input("Enter num1: "))
opr = input("Enter opr: ")
num2 = int(input("Enter num2: "))

ans = 0

if opr == '+':
    print(f"{num1} {opr} {num2} = {num1 + num2}")
elif opr == '-':
    print(f"{num1} {opr} {num2} = {num1 - num2}")
elif opr == '*':
    print(f"{num1} {opr} {num2} = {num1 * num2}")
elif opr == '/':
    print(f"{num1} {opr} {num2} = {round(num1 / num2, 2)}")
else:
    print("Error: invalid operator")
```
