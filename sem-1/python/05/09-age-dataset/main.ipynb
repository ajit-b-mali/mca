```python
import numpy as np

STUDENT_CNT = 10

print(f"Enter the age dataset of {STUDENT_CNT} students")

students = []
for i in range(STUDENT_CNT):
    temp = int(input(f"student {i + 1}: "))
    students.append(temp)

print(f"1  : Students average age: {np.average(students)}")
print(f"2  : Students mean age: {np.mean(students)}")
print(f"3  : Students median age: {np.median(students)}")
print(f"4.a: Count of Students With highest age {max(students)}: {students.count(max(students))}")
print(f"4.b: Count of Students With lowest age {min(students)}: {students.count(min(students))}")
```
